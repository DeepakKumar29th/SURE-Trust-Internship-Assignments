# Question 1 - Food Delivery Order System

## Step 1 - Customer Class

```python

class Customer:

    # Constructor
    def __init__(self, name, wallet_balance):

        self.name = name

        # Private variable
        self.__wallet_balance = wallet_balance

    # Get wallet balance
    def get_wallet_balance(self):

        return self.__wallet_balance

    # Payment method
    def pay_from_wallet(self, amount):

        if amount <= 0:

            print("Invalid payment amount.")
            return False

        elif amount <= self.__wallet_balance:

            self.__wallet_balance -= amount

            print("Payment successful.")
            return True

        else:

            print("Payment failed. Not enough wallet balance.")
            return False

    # Display customer details
    def display_customer_details(self):

        print("\nCustomer Details")
        print("Customer Name:", self.name)
        print("Wallet Balance:", self.__wallet_balance)
```

## Step 2 - FoodItem Class

```python

class FoodItem:

    # Constructor
    def __init__(self, item_name, price, stock):

        self.item_name = item_name

        # Private variables
        self.__price = price
        self.__stock = stock

    # Get item price
    def get_price(self):

        return self.__price

    # Get stock quantity
    def get_stock(self):

        return self.__stock

    # Reduce stock
    def reduce_stock(self, quantity):

        if quantity <= 0:

            return False

        elif quantity <= self.__stock:

            self.__stock -= quantity

            print("Stock reduced successfully.")
            return True

        else:

            print("Food item out of stock.")
            return False

    # Display food item details
    def display_item_details(self):

        print("\nFood Item Details")
        print("Item Name:", self.item_name)
        print("Price:", self.__price)
        print("Available Stock:", self.__stock)
```

## Step 3 - FoodOrder Class

```python

class FoodOrder:

    # Constructor
    def __init__(self, customer, food_item, quantity):

        # Object linking
        self.customer = customer
        self.food_item = food_item

        self.quantity = quantity

        self.total_amount = 0

        self.order_status = "Pending"

    # Place order
    def place_order(self):

        try:

            quantity = int(self.quantity)

            # Check zero quantity
            if quantity <= 0:

                self.order_status = (
                    "Failed - Invalid Quantity"
                )

                return

            # Check stock availability
            if quantity > self.food_item.get_stock():

                self.order_status = (
                    "Failed - Food Quantity Not Available"
                )

                return

            # Calculate total amount
            self.total_amount = (
                self.food_item.get_price() * quantity
            )

            # Wallet payment
            payment_successful = (
                self.customer.pay_from_wallet(
                    self.total_amount
                )
            )

            # If payment successful
            if payment_successful:

                self.food_item.reduce_stock(quantity)

                self.order_status = "Order Placed"

            else:

                self.order_status = (
                    "Failed - Low Wallet Balance"
                )

        except ValueError:

            self.order_status = (
                "Failed - Quantity Is Not A Number"
            )

    # Display order details
    def display_order_details(self):

        print("\nOrder Details")
        print("Customer Name:", self.customer.name)
        print("Food Item Name:", self.food_item.item_name)
        print("Quantity:", self.quantity)
        print("Total Amount:", self.total_amount)
        print("Order Status:", self.order_status)
```

## Step 4 - OrderFileManager Class

```python

class OrderFileManager:

    # Constructor
    def __init__(self, file_name):

        self.file_name = file_name

    # Save order into text file
    def save_order(self, order):

        with open(self.file_name, "a") as file:

            file.write(
                "Customer Name: "
                + order.customer.name + "\n"
            )

            file.write(
                "Food Item Name: "
                + order.food_item.item_name + "\n"
            )

            file.write(
                "Quantity: "
                + str(order.quantity) + "\n"
            )

            file.write(
                "Total Amount: "
                + str(order.total_amount) + "\n"
            )

            file.write(
                "Order Status: "
                + order.order_status + "\n"
            )

            file.write("--------------------------\n")

        print("Order saved successfully.")
```

## Step 5 - Create Objects

```python

# Customer objects
customer1 = Customer("Ravi", 1000)

customer2 = Customer("Priya", 100)

# Food item objects
item1 = FoodItem("Biryani", 300, 10)

item2 = FoodItem("Pizza", 500, 2)

item3 = FoodItem("Burger", 200, 5)
```

# Question 2 - Movie Ticket Booking System

## Step 7 - Movie Class

```python

class Movie:

    # Constructor
    def __init__(self, movie_name, ticket_price, seats):

        self.movie_name = movie_name

        # Private variables
        self.__ticket_price = ticket_price
        self.__seats = seats

    # Get ticket price
    def get_ticket_price(self):

        return self.__ticket_price

    # Get seats
    def get_seats(self):

        return self.__seats

    # Reduce seats
    def reduce_seats(self, quantity):

        if quantity <= 0:

            return False

        elif quantity <= self.__seats:

            self.__seats -= quantity

            return True

        else:

            return False

    # Display movie details
    def display_movie_details(self):

        print("\nMovie Details")
        print("Movie Name:", self.movie_name)
        print("Ticket Price:", self.__ticket_price)
        print("Available Seats:", self.__seats)
```

## Step 8 - Booking Class

```python

class Booking:

    # Constructor
    def __init__(self, customer, movie, quantity):

        self.customer = customer
        self.movie = movie

        self.quantity = quantity

        self.total_amount = 0

        self.booking_status = "Pending"

    # Place booking
    def place_booking(self):

        try:

            quantity = int(self.quantity)

            # Check invalid quantity
            if quantity <= 0:

                self.booking_status = (
                    "Failed - Invalid Quantity"
                )

                return

            # Check seats availability
            if quantity > self.movie.get_seats():

                self.booking_status = (
                    "Failed - Seats Not Available"
                )

                return

            # Calculate total amount
            self.total_amount = (
                self.movie.get_ticket_price()
                * quantity
            )

            # Wallet payment
            payment_successful = (
                self.customer.pay_from_wallet(
                    self.total_amount
                )
            )

            # If payment successful
            if payment_successful:

                self.movie.reduce_seats(quantity)

                self.booking_status = (
                    "Booking Confirmed"
                )

            else:

                self.booking_status = (
                    "Failed - Low Wallet Balance"
                )

        except ValueError:

            self.booking_status = (
                "Failed - Seats Is Not A Number"
            )

    # Display booking details
    def display_booking_details(self):

        print("\nBooking Details")
        print("Customer Name:", self.customer.name)
        print("Movie Name:", self.movie.movie_name)
        print("Seats Booked:", self.quantity)
        print("Total Amount:", self.total_amount)
        print("Booking Status:", self.booking_status)
```

## Step 9 - BookingFileManager Class

```python

class BookingFileManager:

    # Constructor
    def __init__(self, file_name):

        self.file_name = file_name

    # Save booking into file
    def save_booking(self, booking):

        with open(self.file_name, "a") as file:

            file.write(
                "Customer Name: "
                + booking.customer.name + "\n"
            )

            file.write(
                "Movie Name: "
                + booking.movie.movie_name + "\n"
            )

            file.write(
                "Seats Booked: "
                + str(booking.quantity) + "\n"
            )

            file.write(
                "Total Amount: "
                + str(booking.total_amount) + "\n"
            )

            file.write(
                "Booking Status: "
                + booking.booking_status + "\n"
            )

            file.write("--------------------------\n")

        print("Booking saved successfully.")
```
