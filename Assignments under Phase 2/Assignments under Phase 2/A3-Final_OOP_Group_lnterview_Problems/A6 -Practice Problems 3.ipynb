# Problem 3: Movie Ticket Booking and Seat Management System

# CREATING CUSTOMER CLASS

```python
class Customer:

    def __init__(self, name, wallet_balance):
        self.name = name
        self.__wallet_balance = wallet_balance

    def get_wallet_balance(self):
        return self.__wallet_balance

    def pay_from_wallet(self, amount):
        if amount <= self.__wallet_balance:
            self.__wallet_balance -= amount
            return True
        else:
            return False
    def get_discount_percentage(self):
        return 0
```

# creating Special Customer Types

```python
class StudentCustomer(Customer):
    def get_discount_percentage(self):
        return 10

class PremiumCustomer(Customer):
    def get_discount_percentage(self):
        return 20
```

# Creating Movie Class

```python
class Movie:
    def __init__(self, movie_name, ticket_price, seats):
        self.movie_name = movie_name
        self.__ticket_price = ticket_price
        self.__seats = seats

    def get_ticket_price(self):
        return self.__ticket_price

    def get_seats(self):
        return self.__seats

    def reduce_seats(self, quantity):
        if quantity <= self.__seats:
            self.__seats -= quantity
            return True
        else:
            return False

    def display_movie_details(self):
        print("\nMovie Details")
        print("Movie Name:", self.movie_name)
        print("Ticket Price:", self.__ticket_price)
        print("Available Seats:", self.__seats)
```

# Creating Booking Class

```python
class Booking:

    def __init__(self, customer, movie, quantity):
        self.customer = customer
        self.movie = movie
        self.quantity = quantity
        self.total_amount = 0
        self.discount_amount = 0
        self.final_amount = 0
        self.booking_status = "Pending"
        self.failure_reason = "None"

    def place_booking(self):
        try:
            quantity = int(self.quantity)
            if quantity <= 0:
                self.booking_status = "Failed"
                self.failure_reason = "Invalid Seat Quantity"
                return
            if quantity > self.movie.get_seats():
                self.booking_status = "Failed"
                self.failure_reason = "Seats Not Available"
                return
            self.total_amount = (self.movie.get_ticket_price()* quantity)
            discount_percentage = (self.customer.get_discount_percentage())
            self.discount_amount = (self.total_amount* discount_percentage / 100)
            self.final_amount = (self.total_amount- self.discount_amount)
            payment_successful = (self.customer.pay_from_wallet(self.final_amount))

            if not payment_successful:
                self.booking_status = "Failed"
                self.failure_reason = "Low Wallet Balance"
                return
            self.movie.reduce_seats(quantity)
            self.booking_status = "Booking Confirmed"

        except ValueError:
            self.booking_status = "Failed"
            self.failure_reason = ( "Seats Input Is Not A Number")

    def display_booking_details(self):
        print("\nBooking Details")
        print("Customer Name:", self.customer.name)
        print("Movie Name:", self.movie.movie_name)
        print("Seats Booked:", self.quantity)
        print("Total Amount:", self.total_amount)
        print("Discount Amount:", self.discount_amount)
        print("Final Amount:", self.final_amount)
        print("Booking Status:", self.booking_status)
        print("Failure Reason:", self.failure_reason)
```

# Creating Booking File Manager

```python
class BookingFileManager:

    def __init__(self, file_name):
        self.file_name = file_name

    def save_booking(self, booking):
        with open(self.file_name, "a") as file:
            file.write("Customer Name: "+ booking.customer.name + "\n")
            file.write("Movie Name: "+ booking.movie.movie_name + "\n")
            file.write("Seats Booked: "+ str(booking.quantity) + "\n")
            file.write("Final Amount: "+ str(booking.final_amount) + "\n")
            file.write("Booking Status: "+ booking.booking_status + "\n")
            file.write("Failure Reason: "+ booking.failure_reason + "\n")
            file.write("----------------------\n")
    print("Booking saved successfully.")
```

# Creating Report Manager

```python
class ReportManager:

    def generate_report(self, bookings):
        total_tickets_booked = 0
        total_revenue = 0

        successful_bookings = {}
        failed_bookings = {}
        remaining_seats = {}

        for booking in bookings:
            movie_name = booking.movie.movie_name
            remaining_seats[movie_name] = (booking.movie.get_seats())

            if booking.booking_status == ("Booking Confirmed"):
                total_tickets_booked += int(booking.quantity)
                total_revenue += (booking.final_amount)

                if movie_name in successful_bookings:
                    successful_bookings[movie_name] += 1
                else:
                    successful_bookings[movie_name] = 1
            else:
                if movie_name in failed_bookings:
                    failed_bookings[movie_name] += 1
                else:
                    failed_bookings[movie_name] = 1

        print("\n========== MOVIE REPORT ==========")
        print("Total Tickets Booked:",total_tickets_booked)
        print("Total Booking Revenue:",total_revenue)
        print("Movie Wise Successful Bookings:",successful_bookings)
        print("Movie Wise Failed Bookings:",failed_bookings)
        print("Remaining Seats Report:",remaining_seats)
```

# Creating Customers and Movies

```python
customer1 = Customer("Deepak",1000)
customer2 = StudentCustomer("Daya",1000)
customer3 = PremiumCustomer("Hemant",500)

movie1 = Movie("Avengers",250,10)
movie2 = Movie("Interstellar",300,2)

movie1.display_movie_details()
movie2.display_movie_details()
```

# Creating Booking Objects

```python
booking1 = Booking(customer1, movie1, 2)
booking2 = Booking(customer2, movie1, 2)
booking3 = Booking(customer3, movie1, 5)
booking4 = Booking(customer1, movie2, 10)
booking5 = Booking(customer1, movie1, "two")
booking6 = Booking(customer1, movie1, 0)
```

# Placing Bookings

```python
bookings = [booking1,booking2,booking3,booking4,booking5,booking6]

for booking in bookings:
    booking.place_booking()
    booking.display_booking_details()
```

# Saving Bookings into File

```python
file_manager = BookingFileManager("bookings.txt")
for booking in bookings:file_manager.save_booking(booking)
```

# Final Report

```python
report = ReportManager()
report.generate_report(bookings)
```
