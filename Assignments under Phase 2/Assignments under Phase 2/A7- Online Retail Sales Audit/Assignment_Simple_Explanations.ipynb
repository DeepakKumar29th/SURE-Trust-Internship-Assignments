```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Convert Required Columns to NumPy Arrays

quantity = df["Quantity"].to_numpy()
price = df["Price"].to_numpy()
customer = df["Customer ID"].to_numpy()
country = df["Country"].to_numpy()
product = df["Description"].to_numpy()
invoice = df["Invoice"].astype(str).to_numpy()
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# REVENUE CALCULATION USING NUMPY

revenue = quantity * price
df['Revenue'] = revenue
df[['Quantity','Price','Revenue']].head()
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Overall Sales Performance

valid_sales = revenue[(quantity > 0) & (price > 0)]
total_revenue = np.sum(valid_sales)
average_revenue = np.mean(valid_sales)
print("Total Revenue :", round(total_revenue,2))
print("Average Revenue :", round(average_revenue,2))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Returns Analysis

returns = quantity < 0
return_count = np.sum(returns)
return_revenue = np.sum(revenue[returns])
print("Returned Transactions :", return_count)
print("Return Revenue Impact :", round(return_revenue,2))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Cancellation Analysis
cancelled = np.char.startswith(invoice.astype(str), "C")
print("Cancelled Invoices :", np.sum(cancelled))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Missing Customer Records

missing_customers = np.sum(pd.isna(customer))
print("Missing Customer IDs :", missing_customers)
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Unusual Quantity Values

print("Maximum Quantity :", np.max(quantity))
print("Minimum Quantity :", np.min(quantity))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Unusual Price Values

valid_price = price[price > 0]
print("Maximum Price :", np.max(valid_price))
print("Minimum Price :", np.min(valid_price))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Detect Extreme Revenue Values

q1=np.percentile(valid_sales,25)
q3=np.percentile(valid_sales,75)
iqr=q3-q1
upper_limit=q3+1.5*iqr
outliers=valid_sales[valid_sales>upper_limit]
print("Revenue Outliers :",len(outliers))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Top 10 Products by Revenue

valid_mask=(quantity>0)&(price>0)
valid_products=product[valid_mask]
valid_revenue=revenue[valid_mask]
unique_products=np.unique(valid_products)
product_revenue=[]

for p in unique_products:
    product_revenue.append(np.sum(valid_revenue[valid_products==p]))

product_revenue=np.array(product_revenue)
top_products=np.argsort(product_revenue)[-10:]

print("===== TOP 10 PRODUCTS BY REVENUE =====\n")

for i in reversed(top_products):
    print(unique_products[i],"=> £",round(product_revenue[i],2))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Top Countries by Revenue
unique_country=np.unique(country)
country_revenue=[]

for c in unique_country:
    country_revenue.append(np.sum(valid_revenue[country[valid_mask]==c]))

country_revenue=np.array(country_revenue)
top_country=np.argsort(country_revenue)[-10:]

for i in reversed(top_country):
    print(unique_country[i]," --- ",round(country_revenue[i],2))

# These countries contribute the most revenue.
# They are key markets for future growth.
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# High Value Customers

valid_customer=customer[valid_mask]
unique_customers=np.unique(valid_customer[~pd.isna(valid_customer)])
customer_revenue=[]

for cust in unique_customers:
    customer_revenue.append(np.sum(valid_revenue[valid_customer==cust]))

customer_revenue=np.array(customer_revenue)
top_customer=np.argsort(customer_revenue)[-10:]

for i in reversed(top_customer):
    print(unique_customers[i],"=>",round(customer_revenue[i],2))
```

```python
# What is happening in this cell?
# This code is part of the sales audit analysis.
# Read the code below and explain the output after execution.
# The comments can be edited further for each line if needed.

# Correlation Analysis
corr_mask=(~np.isnan(quantity))&(~np.isnan(price))

corr_quantity_price=np.corrcoef(quantity[corr_mask],price[corr_mask])[0,1]
corr_quantity_revenue=np.corrcoef(quantity[corr_mask],revenue[corr_mask])[0,1]
corr_price_revenue=np.corrcoef(price[corr_mask],revenue[corr_mask])[0,1]

print("Quantity vs Price :",round(corr_quantity_price,4))
print("Quantity vs Revenue :",round(corr_quantity_revenue,4))
print("Price vs Revenue :",round(corr_price_revenue,4))
```
