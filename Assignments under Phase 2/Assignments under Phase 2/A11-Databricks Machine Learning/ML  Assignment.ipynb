```python
from pyspark.sql.functions import col, when
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.ml.classification import LogisticRegression
from pyspark.ml import Pipeline
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
import mlflow
import mlflow.spark
```

```python
df = spark.read.csv("/Workspace/Users/s.dk2004th@gmail.com/Databricks ML Assignment/ai4i2020.csv",header=True,inferSchema=True)
display(df)
```

```python
print("Number of Rows :", df.count())
print("Number of Columns :", len(df.columns))
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col, sum
df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).show()
```

```python
# Target Variable Distribution

display(df.groupBy("Machine failure").count())
```

```python
# Removing Unnecessary Columns

df = df.drop("UDI", "Product ID")
```

## Data Preprocessing

```python
# Removinh Data Leakage Columns

df=df.drop("TWF","HDF","PWF","OSF","RNF")
```

```python
print(df.columns)
```

```python
# Convert Type into numeric
indexer=StringIndexer(inputCol="Type",outputCol="Type_Index")
```

```python
# Create feature vector
assembler=VectorAssembler(inputCols=["Type_Index","Air temperature [K]","Process temperature [K]","Rotational speed [rpm]","Torque [Nm]","Tool wear [min]"],outputCol="features")

pipeline=Pipeline(stages=[indexer,assembler])

model=pipeline.fit(df)

prepared_df=model.transform(df)

final_df=prepared_df.select("features",col("Machine failure").alias("label"))

display(final_df)
```

```python
# Train Test Split

train_df, test_df = final_df.randomSplit([0.8,0.2],seed=42)
```

```python
print("Training Records :", train_df.count())
print("Testing Records :", test_df.count())
```

```python
# Train Logistic Regression Model

lr=LogisticRegression(featuresCol="features",labelCol="label")
model=lr.fit(train_df)
```

```python
# # Model Prediction

predictions=model.transform(test_df)

display(predictions.select("label","prediction","probability"))
```

```python
# # Model Evaluation

accuracy=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="accuracy").evaluate(predictions)
precision=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="weightedPrecision").evaluate(predictions)
recall=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="weightedRecall").evaluate(predictions)
f1=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="f1").evaluate(predictions)
```

```python
# # MLflow Experiment Tracking

import mlflow

with mlflow.start_run():
    lr=LogisticRegression(featuresCol="features",labelCol="label")
    model=lr.fit(train_df)
    predictions=model.transform(test_df)

    accuracy=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="accuracy").evaluate(predictions)
    precision=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="weightedPrecision").evaluate(predictions)
    recall=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="weightedRecall").evaluate(predictions)
    f1=MulticlassClassificationEvaluator(labelCol="label",predictionCol="prediction",metricName="f1").evaluate(predictions)

    mlflow.log_metric("Accuracy",accuracy)
    mlflow.log_metric("Precision",precision)
    mlflow.log_metric("Recall",recall)
    mlflow.log_metric("F1 Score",f1)

    print("MLflow metrics logged successfully.")
```

```python
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

```python
# # Maintenance Risk Classification

from pyspark.sql.functions import udf,col
from pyspark.sql.types import StringType

def risk_level(probability):
    p=float(probability[1])
    if p<0.30:
        return "Normal"
    elif p<0.70:
        return "Watch"
    else:
        return "Urgent"

def maintenance_action(probability):
    p=float(probability[1])
    if p<0.30:
        return "Continue Operation"
    elif p<0.70:
        return "Schedule Inspection"
    else:
        return "Immediate Maintenance"

risk_udf=udf(risk_level,StringType())
action_udf=udf(maintenance_action,StringType())

result=predictions.withColumn("Risk_Level",risk_udf(col("probability")))\
                  .withColumn("Maintenance_Action",action_udf(col("probability")))

display(result.select("label","prediction","probability","Risk_Level","Maintenance_Action"))
```

```python
# # Save Results as Delta Table

result.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("predictive_maintenance_result_v2")
```
