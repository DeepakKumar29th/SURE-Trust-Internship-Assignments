## Data Cleaning Process

```python
df['city'] = df['city'].str.strip().str.title()
df['city'].head()
```

```python
df['course_category'] = (df['course_category'].str.strip().str.title())
df['course_category'].head()
```

```python
df['course_fee'] = (
    df['course_fee']
    .astype(str)
    .str.replace('₹','', regex=False)
    .str.replace('Rs.','', regex=False)
    .str.replace('INR','', regex=False)
    .str.replace('.','', regex=False)
    .str.strip()
)


df['course_fee'] = pd.to_numeric(df['course_fee'],errors='coerce')
df[['course_fee']].head()
```

```python
# Handling Missing Student Names
df['student_name'] = (df['student_name'].fillna('Unknown Student'))

# Handling Missing Discount Codes
df['discount_code'] = (df['discount_code'].fillna('No Discount'))

df[['student_name','discount_code']].head(10)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
len(df)
```

## Revenue Analysis


```python
# Creating Gross Fee

df['gross_fee'] = df['course_fee']
df[['course_fee','gross_fee']].head()
```

```python
# Revenue Summary

print("Total Gross Revenue:", df["gross_fee"].sum())
print("Total Valid Revenue:",df["valid_fee_collected"].sum())
```

```python
# Creating Valid Fee Collected

# Revenue counts only when:
# Payment = Paid
# Completion = Completed

df['valid_fee_collected'] = 0
df.loc[(df['payment_status'] == 'Paid') &(df['completion_status'] == 'Completed'),'valid_fee_collected'] = df['gross_fee']

df[['enrollment_id','payment_status','completion_status','gross_fee','valid_fee_collected']].head()
```

```python
# Failed Payment Enrollments

failed_payments = df[df['payment_status'] == 'Failed']
print('Failed Payments:', len(failed_payments))

failed_payments[['enrollment_id','student_name','course_name','gross_fee']].head()
```

```python
# Incomplete Course Enrollments

incomplete_courses = df[df['completion_status'] == 'Incomplete']
print('Incomplete Enrollments:', len(incomplete_courses))

incomplete_courses[['enrollment_id','student_name','course_name']].head()
```

```python
#  Paid But Incomplete

paid_incomplete = df[(df['payment_status'] == 'Paid') &(df['completion_status'] == 'Incomplete')]
print('Paid But Incomplete:', len(paid_incomplete))

paid_incomplete[['student_name','course_name','gross_fee']].head()
```

```python
# Payment Failed OR Course Incomplete

problem_enrollments = df[(df['payment_status'] == 'Failed') |(df['completion_status'] == 'Incomplete')]
print('Problem Enrollments:', len(problem_enrollments))

problem_enrollments.head()
```

```python
# High Value Failed Enrollments Above 5000

high_value_failed = df[(df['payment_status'] == 'Failed') &(df['gross_fee'] > 5000)]
print('High Value Failed Enrollments:', len(high_value_failed))

high_value_failed[['student_name','course_name','gross_fee']].head()
```

```python
# Filter Students From Selected Cities

selected_cities = df[df['city'].isin(['Chennai','Bangalore','Coimbatore'])]
print('Selected City Enrollments:', len(selected_cities))

selected_cities[['student_name','city']].head()
```

```python
# Course Fee Between 3000 and 10000

fee_range = df[(df['gross_fee'] >= 3000) &(df['gross_fee'] <= 10000)]
print('Enrollments Between 3000 and 10000:', len(fee_range))

fee_range[['student_name','gross_fee']].head()
```

```python
# Filter Enrollments That Used a Discount Code

discount_enrollments = df[df['discount_code'] != 'No Discount']
print('Discount Code Enrollments:', len(discount_enrollments))

discount_enrollments[['student_name','discount_code']].head()
```

```python
# Discount Code Enrollments With Zero Revenue

discount_zero_revenue = df[(df['discount_code'] != 'No Discount') &(df['valid_fee_collected'] == 0)]
print('Discount But No Revenue:', len(discount_zero_revenue))

discount_zero_revenue[['student_name','discount_code','valid_fee_collected']].head()
```
