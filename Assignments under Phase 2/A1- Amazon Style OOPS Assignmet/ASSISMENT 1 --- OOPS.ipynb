# Step 1 - Customer Class

This class stores customer information


```python
class Customer:
    def __init__(self, customer_id, name, email, phone, city):
        self.customer_id = customer_id
        self.name = name
        self.email = email
        self.phone = phone
        self.city = city

    def display_customer_details(self):  # Display customer details
        print("\nCustomer Details")
        print("Customer ID:", self.customer_id)
        print("Name:", self.name)
        print("Email:", self.email)
        print("Phone:", self.phone)
        print("City:", self.city)

    def update_phone(self, new_phone):     # Update phone number
        self.phone = new_phone
        print("\nPhone number updated successfully")

    def update_city(self, new_city):     # Update city
        self.city = new_city
        print("\nCity updated successfully")
```


# Step 2 - Product Class

This class stores product information


```python
class Product:
    def __init__(self, product_id, product_name, category, price, stock_quantity):
        self.product_id = product_id
        self.product_name = product_name
        self.category = category
        self.price = price
        self.stock_quantity = stock_quantity

    def display_product_details(self):     # Display product details
        print("\nProduct Details")
        print("Product ID:", self.product_id)
        print("Product Name:", self.product_name)
        print("Category:", self.category)
        print("Price:", self.price)
        print("Stock:", self.stock_quantity)

    def update_price(self, new_price):
        if new_price > 0:             # Update price
            self.price = new_price
            print("\nPrice updated successfully")
        else:
            print("\nInvalid price")

    def is_stock_available(self, quantity):     # Check stock availability
        if quantity <= self.stock_quantity:
            return True
        else:
            return False
    def reduce_stock(self, quantity):     # Reduce stock after successful order
        self.stock_quantity = self.stock_quantity - quantity
```


# Step 3 - Order Class

This class connects Customer and Product


```python
class Order:
    def __init__(self, order_id, customer, product, quantity):
        self.order_id = order_id
        self.customer = customer
        self.product = product
        self.quantity = quantity
        self.total_amount = 0
        self.order_status = "Pending"

    def calculate_total(self):     # Calculate total amount
        self.total_amount = self.product.price * self.quantity

    def place_order(self):     # Place order
        if self.product.is_stock_available(self.quantity):
            self.calculate_total()
            self.product.reduce_stock(self.quantity)
            self.order_status = "Order Placed"
            print("\nOrder placed successfully")

        else:
            self.order_status = "Order Failed"
            print("\nOrder failed because stock is not available")

    def display_order_details(self):     # Display order details
        print("\nOrder Details")
        print("Order ID:", self.order_id)
        print("Customer Name:", self.customer.name)
        print("Product Name:", self.product.product_name)
        print("Quantity:", self.quantity)
        print("Total Amount:", self.total_amount)
        print("Order Status:", self.order_status)
```


# Step 4 - Invoice Class

This class connects with Order


```python

class Invoice:

    def __init__(self, invoice_id, order, gst_percentage):

        self.invoice_id = invoice_id
        self.order = order
        self.gst_percentage = gst_percentage

        self.gst_amount = 0
        self.final_amount = 0

    def calculate_gst(self):     # Calculate GST
        self.gst_amount = (self.order.total_amount * self.gst_percentage) / 100
    
    def calculate_final_amount(self):
        self.final_amount = self.order.total_amount + self.gst_amount

    def display_invoice(self):     # Display invoice
        if self.order.order_status == "Order Placed":           # Invoice only for successful orders

            self.calculate_gst()
            self.calculate_final_amount()

            print("\nInvoice Details")
            print("Invoice ID:", self.invoice_id)
            print("Customer Name:", self.order.customer.name)
            print("Product Name:", self.order.product.product_name)
            print("Total Amount:", self.order.total_amount)
            print("GST Amount:", self.gst_amount)
            print("Final Amount:", self.final_amount)

        else:

            print("\nInvoice cannot be generated")
```


# Step 5 - Creating Customer Objects
